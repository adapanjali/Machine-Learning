```python
import nltk
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize # Each sentence is separated. 

corpus = "I am Anjali. I like books. This is Anjali's life!"
sentences = sent_tokenize(corpus)
print(sentences)
```

```python
from nltk.tokenize import word_tokenize # Sometimes punctuation marks are not separated. 

words = word_tokenize(corpus)
print(words)
```

```python
from nltk.tokenize import wordpunct_tokenize # Each punctuation mark is separated. 

words2 = wordpunct_tokenize(corpus)
print(words2)
```

```python
from nltk.tokenize import TreebankWordTokenizer # Seems like full stop is intact with the connecting word.
# But last punctuation mark will be separated. 
tbw_tokenizer = TreebankWordTokenizer()
words3 = tbw_tokenizer.tokenize(corpus)
print(words3)
```
