```python
import nltk
```

```python
from nltk.stem import PorterStemmer
stemming = PorterStemmer()

words = ['eating', 'eats', 'ate', 'running', 'runs', 'ran', 'writing', 'writes', 'wrote', 'written', 'eaten', 'history', 'congratulations']

for word in words:
     print(word + "------>" + stemming.stem(word))
```
