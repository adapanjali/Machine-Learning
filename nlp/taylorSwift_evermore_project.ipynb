```python
import requests
from bs4 import BeautifulSoup

def get_azlyrics(url):
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36"
    } 
    res = requests.get(url, headers=headers) # Requests returns a status code
    if res.status_code != 200: ## Status Code 200 means all is OK
        raise Exception(f"Failed to fetch page: {res.status_code}")

    # res.text contains the whole HTML code of the webpage that we are trying to parse
    # parsing is similar to translator - here we are using the html parser
    soup = BeautifulSoup(res.text, "html.parser")

    # Lyrics on AZLyrics are inside the 2nd div after all <div class="ringtone">
    # YOU HAVE TO CHECK THIS PART YOURSELF
    divs = soup.find_all("div")

    for i in range(len(divs)):
        if divs[i].get("class") is None and divs[i].text.strip() != "":
            lyrics = divs[i].get_text(separator="\n").strip()
            return lyrics

    raise Exception("Could not find lyrics")

url = "https://www.azlyrics.com/lyrics/taylorswift/willow.html"
willow_lyrics = get_azlyrics(url=url)
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from selenium_stealth import stealth
from webdriver_manager.chrome import ChromeDriverManager
from bs4 import BeautifulSoup
import time

def get_lyrics_azlyrics(url):
    options = Options()
    options.add_argument("--headless")  # Needed for Codespaces
    options.add_argument("--disable-gpu")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    options.add_argument("user-agent=Mozilla/5.0")

    # Setup WebDriver
    driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

    # Apply stealth to bypass bot detection
    stealth(driver,
            languages=["en-US", "en"],
            vendor="Google Inc.",
            platform="Win32",
            webgl_vendor="Intel Inc.",
            renderer="Intel Iris OpenGL Engine",
            fix_hairline=True)

    driver.get(url)

    # Mimic scrolling and wait for content
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(5)

    soup = BeautifulSoup(driver.page_source, "html.parser")
    print(soup.prettify()[:1500])  # Show the first 1500 characters of the HTML

    driver.quit()

    # Extract lyrics: find all divs with no attributes, get the second one
    divs = soup.find_all("div")
    raw_divs = [div for div in divs if div.attrs == {} and len(div.get_text(strip=True)) > 100]

    return raw_divs[0].get_text(separator="\n").strip() if raw_divs else "Lyrics not found."

# Example usage
url = "https://www.azlyrics.com/lyrics/taylorswift/willow.html"
lyrics = get_lyrics_azlyrics(url)
print(lyrics)
```

```python
import requests
from bs4 import BeautifulSoup

def get_genius_lyrics(url):
    headers = {"User-Agent": "Mozilla/5.0"}
    response = requests.get(url, headers=headers)

    soup = BeautifulSoup(response.text, "html.parser")
    lyrics_div = soup.find("div", {"data-lyrics-container": "true"})
    if not lyrics_div:
        return "Lyrics not found."

    lyrics = "\n".join([line.get_text() for line in lyrics_div.find_all("p")])
    return lyrics.strip()

url = "https://genius.com/Taylor-swift-willow-lyrics"
lyrics = get_genius_lyrics(url)
print(lyrics)
```

```python
import requests
from bs4 import BeautifulSoup

def search_genius(song_name):
    base_url = "https://genius.com"
    headers = {"User-Agent": "Mozilla/5.0"}
    search_url = f"https://genius.com/api/search/multi?per_page=1&q={song_name.replace(' ', '%20')}"

    res = requests.get(search_url, headers=headers).json()

    # Find song hit from API response
    sections = res.get("response", {}).get("sections", [])
    for section in sections:
        if section["type"] == "song":
            hits = section["hits"]
            if hits:
                return hits[0]["result"]["url"]
    return None

def get_genius_lyrics(url):
    headers = {"User-Agent": "Mozilla/5.0"}
    response = requests.get(url, headers=headers)
    soup = BeautifulSoup(response.text, "html.parser")

    # Lyrics are inside divs with this attribute
    lyrics_divs = soup.find_all("div", {"data-lyrics-container": "true"})
    if not lyrics_divs:
        return "Lyrics not found."

    lyrics = "\n".join(p.get_text(strip=True) for div in lyrics_divs for p in div.find_all("p"))
    return lyrics.strip()

# 🎯 Enter your song here
query = "Style Taylor Swift"

print(f"🔍 Searching Genius for: {query}")
song_url = search_genius(query)

if song_url:
    print(f"✅ Found URL: {song_url}")
    lyrics = get_genius_lyrics(song_url)
    print("\n🎵 Lyrics:\n")
    print(lyrics)
else:
    print("❌ No song found on Genius.")
```

```python
import requests
from bs4 import BeautifulSoup

def search_genius_html(song_name):
    headers = {"User-Agent": "Mozilla/5.0"}
    search_url = f"https://genius.com/search?q={song_name.replace(' ', '%20')}"

    response = requests.get(search_url, headers=headers)
    if response.status_code != 200:
        print(f"❌ Genius search failed: {response.status_code}")
        return None

    soup = BeautifulSoup(response.text, "html.parser")
    first_result = soup.select_one("a[href^='https://genius.com/'][data-testid='search_result']")

    if first_result:
        return first_result["href"]
    return None

def get_genius_lyrics(url):
    headers = {"User-Agent": "Mozilla/5.0"}
    response = requests.get(url, headers=headers)
    soup = BeautifulSoup(response.text, "html.parser")

    lyrics_divs = soup.find_all("div", {"data-lyrics-container": "true"})
    if not lyrics_divs:
        return "Lyrics not found."

    lyrics = "\n".join(p.get_text(strip=True) for div in lyrics_divs for p in div.find_all("p"))
    return lyrics.strip()

# 🔍 Try it
query = "Style Taylor Swift"
print(f"🔍 Searching Genius for: {query}")
song_url = search_genius_html(query)

if song_url:
    print(f"✅ Found URL: {song_url}")
    lyrics = get_genius_lyrics(song_url)
    print("\n🎵 Lyrics:\n")
    print(lyrics)
else:
    print("❌ No song found on Genius.")
```
